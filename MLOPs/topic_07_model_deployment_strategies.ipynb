# 🚀 Topic 07: Model Deployment Strategies & Traffic Routing

## 1. Deployment Strategies Overview

| Strategy | Description | Zero Downtime? | Rollback Speed | Risk Level |
| :--- | :--- | :--- | :--- | :--- |
| **Recreate** | Terminate old version, boot new version | ❌ No | Slow | 🔴 High |
| **Blue-Green** | Duplicate environment; switch router 100% at once | ✅ Yes | Instant | 🟡 Medium |
| **Canary** | Route small % of traffic to new version; ramp up | ✅ Yes | Instant | 🟢 Low |
| **Shadow** | Duplicate request traffic to v2 without user impact | ✅ Yes | N/A | 🟢 Zero |

---

## 2. Hands-on: Simulating Canary Routing & Traffic Splitting


```python
import random
import numpy as np

class ModelV1_Baseline:
    def predict(self, x):
        return 0

class ModelV2_Canary:
    def predict(self, x):
        return 1

class CanaryRouter:
    def __init__(self, canary_weight=0.10):
        self.v1 = ModelV1_Baseline()
        self.v2 = ModelV2_Canary()
        self.canary_weight = canary_weight
        self.metrics = {"v1_requests": 0, "v2_requests": 0}

    def route_request(self, input_data):
        if random.random() < self.canary_weight:
            self.metrics["v2_requests"] += 1
            return {"version": "v2_canary", "result": self.v2.predict(input_data)}
        else:
            self.metrics["v1_requests"] += 1
            return {"version": "v1_baseline", "result": self.v1.predict(input_data)}

router = CanaryRouter(canary_weight=0.15)

for _ in range(1000):
    router.route_request(input_data=[0.5, 0.9])

print("📊 Traffic Routing Results:")
print(f"   Baseline v1 Requests: {router.metrics['v1_requests']} ({router.metrics['v1_requests']/10:.1f}%)")
print(f"   Canary   v2 Requests: {router.metrics['v2_requests']} ({router.metrics['v2_requests']/10:.1f}%)")
```
