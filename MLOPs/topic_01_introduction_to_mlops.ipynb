# 📖 Topic 01: Introduction to MLOps & System Architecture

## 1. What is MLOps?
**MLOps (Machine Learning Operations)** is a set of practices that combines **Machine Learning**, **DevOps**, and **Data Engineering** to reliably and efficiently deploy and maintain ML models in production.

### Why standard DevOps is NOT enough for ML:
- **Code vs Data + Code:** Software engineering versions code. ML systems depend on Code, Data, and Model Weights.
- **Silent Degradation:** Software crashes with stack traces; ML models decay silently as real-world data shifts.
- **Experimental Nature:** ML code development is iterative and non-linear.

---

## 2. MLOps Maturity Levels (Google Framework)

```
Level 0: Manual Process
[ Data Extraction ] -> [ Model Training ] -> [ Manual Hand-off ] -> [ Deploy API ]
(No automated monitoring, manual retrains, silent drift)

Level 1: ML Pipeline Automation
[ Raw Data ] -> [ Automated Pipeline: Prep -> Train -> Eval ] -> [ Model Registry ] -> [ Deployment ]
(Continuous Training (CT) triggered by new data or drift)

Level 2: CI/CD Pipeline Automation
[ Code Repo ] -> [ CI/CD Automated Testing & Build ] -> [ Deploy ML Pipeline ] -> [ Automated Retrain & Serve ]
(Full automation of ML engineering, training, and deployment)
```

## 3. Production ML Directory Structure

Let's generate a standardized production ML project layout.


````python
import os

def create_mlops_structure(base_dir="my_mlops_project"):
    directories = [
        "config",
        "data/raw",
        "data/processed",
        "src/data_prep",
        "src/models",
        "src/pipelines",
        "src/api",
        "tests",
        "notebooks",
        "artifacts"
    ]
    
    for dir_path in directories:
        full_path = os.path.join(base_dir, dir_path)
        os.makedirs(full_path, exist_ok=True)
        if dir_path.startswith("src"):
            with open(os.path.join(full_path, "__init__.py"), "w") as f:
                pass
                
    config_content = "# Production Config\nmodel:\n  name: random_forest\n  n_estimators: 100\n  max_depth: 5\ndata:\n  raw_path: data/raw/dataset.csv\n  processed_path: data/processed/features.csv\n"
    with open(os.path.join(base_dir, "config", "config.yaml"), "w") as f:
        f.write(config_content)
        
    print(f"✅ Generated MLOps Project Structure at: '{base_dir}/'")

create_mlops_structure()
````
