## Hands-on Unified Capstone Execution


```python
import hashlib
import numpy as np
import pandas as pd
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import accuracy_score
from scipy.stats import ks_2samp

print("==================================================")
print("🚀 STARTING END-TO-END MLOPS CAPSTONE PIPELINE")
print("==================================================")

print("\n--- 1. Data Ingestion & Version Control ---")
X_raw, y_raw = make_classification(n_samples=1000, n_features=5, random_state=42)
raw_bytes = X_raw.tobytes()
data_hash = hashlib.md5(raw_bytes).hexdigest()
print(f"Data Loaded. Size: {X_raw.shape} | Version MD5 Hash: {data_hash[:12]}")

X_train, X_test, y_train, y_test = train_test_split(X_raw, y_raw, test_size=0.2, random_state=42)

print("\n--- 2. Model Training & Validation ---")
model = GradientBoostingClassifier(n_estimators=50, random_state=42)
model.fit(X_train, y_train)
acc = accuracy_score(y_test, model.predict(X_test))
print(f"Model Training Complete. Test Accuracy: {acc:.4f}")

print("\n--- 3. Metric Gatekeeper Check ---")
GATE_THRESHOLD = 0.80
if acc >= GATE_THRESHOLD:
    print(f"✅ PASSED: Accuracy ({acc:.4f}) >= Threshold ({GATE_THRESHOLD}). Approved for production!")
else:
    raise ValueError("❌ REJECTED: Model accuracy below deployment threshold!")

print("\n--- 4. Production Drift Monitoring ---")
prod_X = np.random.normal(loc=1.2, scale=1.0, size=(200, 5))
ks_stat, p_val = ks_2samp(X_train[:, 0], prod_X[:, 0])

print(f"Feature 0 KS Stat: {ks_stat:.4f} | p-value: {p_val:.4e}")
if p_val < 0.05:
    print("⚠️ DRIFT ALERT DETECTED! Triggering automated retraining pipeline...")
else:
    print("✅ Production inference healthy. No drift detected.")

print("\n==================================================")
print("🎉 CAPSTONE PIPELINE EXECUTED SUCCESSFULLY!")
print("==================================================")
```
