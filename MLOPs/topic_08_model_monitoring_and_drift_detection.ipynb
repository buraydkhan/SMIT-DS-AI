# 📊 Topic 08: Model Monitoring & Statistical Drift Detection

## 1. Types of Drift
- **Data Drift (Covariate Shift):** Features distribution $P(X)$ changes over time (e.g. user demographic shifts).
- **Concept Drift:** Relationship between features and target $P(Y|X)$ changes (e.g. macro-economic shift changing credit risk scores).

---

## 2. Hands-on: Statistical Drift Detection (Kolmogorov-Smirnov Test & PSI)


```python
import numpy as np
from scipy.stats import ks_2samp

def detect_ks_drift(reference_data, current_data, alpha=0.05):
    ks_stat, p_value = ks_2samp(reference_data, current_data)
    has_drift = p_value < alpha
    print(f"KS Statistic: {ks_stat:.4f} | p-value: {p_value:.4e}")
    if has_drift:
        print("⚠️ DRIFT DETECTED! Production distribution significantly deviates from training data.")
    else:
        print("✅ No Drift Detected. Production data matches baseline.")
    return has_drift

np.random.seed(42)
reference_dist = np.random.normal(0, 1, size=1000)

prod_normal = np.random.normal(0.02, 1.01, size=500)
print("--- Scenario A: Normal Production Traffic ---")
detect_ks_drift(reference_dist, prod_normal)

prod_drifted = np.random.normal(0.8, 1.2, size=500)
print("\n--- Scenario B: Shifted Production Traffic ---")
detect_ks_drift(reference_dist, prod_drifted)
```
