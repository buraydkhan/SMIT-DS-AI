### 1.1 The Artificial Neuron (Perceptron)
An Artificial Neuron takes inputs $x_1, x_2, \dots, x_d$, multiplies each by a weight $w_i$, adds a bias $b$, and passes the sum through an activation function $\sigma$:

$$z = (w_1 x_1 + w_2 x_2 + \dots + w_d x_d) + b = W^T X + b$$
$$\hat{y} = \sigma(z)$$

- **Linear Part ($z$)**: Combines inputs.
- **Activation Function ($\sigma$)**: Adds **non-linearity** so the network can learn non-linear patterns (like curves and complex shapes).


```python
# Define Activation Functions & Their Derivatives
z = np.linspace(-5, 5, 200)

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def sigmoid_grad(z):
    s = sigmoid(z)
    return s * (1 - s)

def relu(z):
    return np.maximum(0, z)

def relu_grad(z):
    return (z > 0).astype(float)

def leaky_relu(z, alpha=0.1):
    return np.where(z > 0, z, alpha * z)

def leaky_relu_grad(z, alpha=0.1):
    return np.where(z > 0, 1.0, alpha)

# Plot Activation Functions & Derivatives
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Sigmoid
axes[0].plot(z, sigmoid(z), label="Sigmoid", color="blue", lw=2)
axes[0].plot(z, sigmoid_grad(z), label="Sigmoid Gradient", color="blue", linestyle="--", lw=2)
axes[0].set_title("1. Sigmoid Function (0 to 1)")
axes[0].grid(True, alpha=0.3)
axes[0].legend()

# ReLU
axes[1].plot(z, relu(z), label="ReLU", color="green", lw=2)
axes[1].plot(z, relu_grad(z), label="ReLU Gradient", color="green", linestyle="--", lw=2)
axes[1].set_title("2. ReLU (Rectified Linear Unit)")
axes[1].grid(True, alpha=0.3)
axes[1].legend()

# Leaky ReLU
axes[2].plot(z, leaky_relu(z), label="Leaky ReLU", color="purple", lw=2)
axes[2].plot(z, leaky_relu_grad(z), label="Leaky ReLU Gradient", color="purple", linestyle="--", lw=2)
axes[2].set_title("3. Leaky ReLU")
axes[2].grid(True, alpha=0.3)
axes[2].legend()

plt.tight_layout()
plt.show()
```

### 1.2 Quick Summary & Cheat-Sheet
- **Sigmoid**: Outputs values between $0$ and $1$. Used for binary classification output layers.
- **ReLU**: Most popular for hidden layers! Extremely fast and avoids vanishing gradients for positive inputs.
- **Leaky ReLU**: Fixes the "Dying ReLU" issue by allowing a small gradient for negative numbers.
- **Softmax**: Converts a vector of numbers into a probability distribution that sums to $1$ (used for multi-class classification).
