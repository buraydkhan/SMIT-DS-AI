### 2.1 Loss Functions Explained Simply

1. **Mean Squared Error (MSE)** — Used for **Regression** (predicting numbers):
   $$\mathcal{L} = \frac{1}{N} \sum_{i=1}^N (y_i - \hat{y}_i)^2$$

2. **Binary Cross-Entropy (BCE)** — Used for **Binary Classification** (0 or 1):
   $$\mathcal{L} = -\frac{1}{N} \sum_{i=1}^N \left[ y_i \log(\hat{y}_i) + (1 - y_i) \log(1 - \hat{y}_i) \right]$$

3. **Categorical Cross-Entropy (CCE)** — Used for **Multi-Class Classification** (Cat, Dog, Bird):
   $$\mathcal{L} = -\sum_{c=1}^C y_c \log(\hat{y}_c)$$


```python
# Simulating Gradient Descent Optimization algorithms

# Objective Function: f(w) = w^2 (Minimum is at w = 0)
def f(w):
    return w ** 2

def grad_f(w):
    return 2 * w

# 1. Vanilla SGD
w_sgd = 4.0
lr = 0.2
history_sgd = [w_sgd]
for _ in range(15):
    w_sgd = w_sgd - lr * grad_f(w_sgd)
    history_sgd.append(w_sgd)

# 2. SGD with Momentum
w_mom = 4.0
v = 0.0
beta = 0.8
history_mom = [w_mom]
for _ in range(15):
    v = beta * v + (1 - beta) * grad_f(w_mom)
    w_mom = w_mom - lr * v
    history_mom.append(w_mom)

# Plot Optimization Trajectories
w_vals = np.linspace(-4.5, 4.5, 100)
plt.figure(figsize=(10, 5))
plt.plot(w_vals, f(w_vals), color="gray", linestyle="--", label="Loss Curve: f(w) = w^2")
plt.plot(history_sgd, [f(w) for w in history_sgd], "o-", color="red", label="Vanilla SGD")
plt.plot(history_mom, [f(w) for w in history_mom], "s-", color="blue", label="SGD + Momentum")

plt.title("Optimization Trajectory: SGD vs SGD + Momentum")
plt.xlabel("Parameter w")
plt.ylabel("Loss f(w)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

### 2.2 Optimizer Comparison
- **SGD (Stochastic Gradient Descent)**: Simple weight update step $\theta \leftarrow \theta - \eta \nabla \mathcal{L}$.
- **Momentum**: Accelerates SGD in the right direction like a ball rolling down a hill.
- **Adam (Adaptive Moment Estimation)**: The gold standard! Automatically adjusts learning rates for each parameter.
