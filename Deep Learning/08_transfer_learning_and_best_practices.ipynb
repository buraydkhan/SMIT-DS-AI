In this final notebook, we look at real-world Deep Learning workflows:
1. **Transfer Learning**: Utilizing pre-trained models (e.g. ResNet) for custom computer vision tasks.
2. **Best Practices Checklist**:
   - Data scaling & normalization
   - Regularization (Dropout, BatchNorm, Weight Decay)
   - Early Stopping & Learning Rate Schedulers
   - Debugging Neural Networks step-by-step


```python
import torch
import torch.nn as nn

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Transfer Learning Example using PyTorch Torchvision
try:
    import torchvision.models as models
    resnet = models.resnet18(weights=models.ResNet18_Weights.DEFAULT)
    
    # 1. Freeze feature extraction backbone
    for param in resnet.parameters():
        param.requires_grad = False
        
    # 2. Replace final fully-connected classifier for custom 5-class task
    in_features = resnet.fc.in_features
    resnet.fc = nn.Linear(in_features, 5)
    resnet = resnet.to(device)
    
    print("ResNet-18 modified for Transfer Learning:")
    print(f"New Classifier Head: {resnet.fc}")
    trainable_params = sum(p.numel() for p in resnet.parameters() if p.requires_grad)
    print(f"Number of trainable parameters: {trainable_params}")
except Exception as e:
    print(f"Torchvision note: {e}")
```

### 🎯 Deep Learning Debugging & Success Checklist
1. **Start Small**: Overfit a tiny batch of 5-10 images/samples first. If your network cannot reach ~0 loss on 10 samples, fix bugs in code/data before scaling up!
2. **Check Learning Rates**: Standard starting learning rate for Adam is `1e-3` or `3e-4`.
3. **Normalize Inputs**: Always scale inputs (e.g. subtract mean, divide by standard deviation).
4. **Use Dropout & BatchNorm**: Reduces overfitting and speeds up model convergence.
5. **Monitor Validation Loss**: Stop training when Validation Loss starts increasing while Training Loss keeps decreasing (Overfitting).
