In this notebook, we cover the architecture driving modern AI (ChatGPT, Claude, Gemini): **Transformers & Self-Attention**!

Topics covered:
1. **Self-Attention Intuition**: Query ($Q$), Key ($K$), Value ($V$).
2. **Scaled Dot-Product Attention Math**:
   $$\text{Attention}(Q, K, V) = \text{softmax}\left( \frac{QK^T}{\sqrt{d_k}} \right) V$$
3. Implementing Self-Attention in PyTorch & visualizing attention heatmaps.


```python
import torch
import torch.nn as nn
import numpy as np
import matplotlib.pyplot as plt

class ScaledDotProductSelfAttention(nn.Module):
    def __init__(self, d_model):
        super(ScaledDotProductSelfAttention, self).__init__()
        self.d_k = d_model
        
    def forward(self, Q, K, V):
        # Shapes: [Batch, Seq_Len, d_model]
        scores = torch.matmul(Q, K.transpose(-2, -1)) / np.sqrt(self.d_k)
        attn_weights = torch.softmax(scores, dim=-1)
        output = torch.matmul(attn_weights, V)
        return output, attn_weights

# Instantiate Self-Attention
batch_size, seq_len, d_model = 1, 6, 16
Q = K = V = torch.randn(batch_size, seq_len, d_model)

attn_layer = ScaledDotProductSelfAttention(d_model)
output, weights = attn_layer(Q, K, V)

print(f"Input Shape: {Q.shape}")
print(f"Self-Attention Output Shape: {output.shape}")
print(f"Attention Weights Matrix Shape: {weights.shape}")

# Visualize Token-to-Token Attention Weights Matrix
plt.figure(figsize=(6, 5))
plt.imshow(weights[0].detach().numpy(), cmap="Blues")
plt.colorbar(label="Attention Weight")
plt.title("Self-Attention Matrix (Token Relationships)")
plt.xlabel("Key Tokens")
plt.ylabel("Query Tokens")
plt.show()
```
