## 1. The 5 V's of Big Data
- **Volume:** Scale of data (terabytes to petabytes).
- **Velocity:** Rate of incoming data generation (streaming logs, IoT sensors).
- **Variety:** Structured (SQL), Semi-structured (JSON, XML), Unstructured (images, video, text).
- **Veracity:** Data quality, noise, and trustworthiness.
- **Value:** Business insights extracted from data.

## 2. CAP Theorem & Distributed Systems

```
              Consistency (C)
                 /       \
                /         \
               /   System  \
              /   Tradeoffs \
             /               \
Availability (A) --------- Partition Tolerance (P)
```
- **Consistency:** Every read receives the most recent write or an error.
- **Availability:** Every request receives a non-error response without guarantee that it contains the latest write.
- **Partition Tolerance:** System continues to operate despite network partition dropouts.

## 3. Hands-on: Building a Python MapReduce Engine from Scratch

MapReduce is the foundational distributed paradigm that preceded Apache Spark.
- **Map Stage:** Transforms input key-value pairs into intermediate key-value pairs.
- **Shuffle/Sort Stage:** Groups all intermediate values associated with the same key.
- **Reduce Stage:** Combines grouped values to produce final aggregates.


````python
from collections import defaultdict
import itertools

# Sample Unstructured Document Corpus
documents = [
    "big data apache spark cloud computing",
    "apache spark distributed computing spark data",
    "cloud computing aws s3 big data spark"
]

# 1. MAPPER FUNCTION
def mapper(document):
    words = document.split()
    return [(word, 1) for word in words]

# 2. SHUFFLE & SORT FUNCTION
def shuffle_and_sort(mapped_results):
    grouped_data = defaultdict(list)
    for key, value in itertools.chain.from_iterable(mapped_results):
        grouped_data[key].append(value)
    return grouped_data

# 3. REDUCER FUNCTION
def reducer(key, values):
    return (key, sum(values))

# --- MAPREDUCE EXECUTION PIPELINE ---
print("--- 1. Mapping Phase ---")
mapped_output = [mapper(doc) for doc in documents]
print("Mapped Output Sample:", mapped_output[0])

print("\n--- 2. Shuffle & Sort Phase ---")
grouped_output = shuffle_and_sort(mapped_output)
print("Grouped Output ('spark'):", grouped_output['spark'])

print("\n--- 3. Reduction Phase ---")
reduced_output = [reducer(key, values) for key, values in grouped_output.items()]
print("Final Word Counts:")
for word, count in sorted(reduced_output, key=lambda x: x[1], reverse=True):
    print(f"   {word:15s}: {count}")
````
