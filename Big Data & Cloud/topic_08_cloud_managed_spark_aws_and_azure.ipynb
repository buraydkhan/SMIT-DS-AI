## 2. AWS Managed Services Comparison
- **AWS EMR (Elastic MapReduce):** Provisioned EC2 cluster running Spark, Hadoop, Presto. Best for predictable, heavy workloads.
- **AWS Glue:** Serverless Spark environment. Automatically scales workers up and down per ETL job. Pay strictly per second of job execution (DPUs).
- **Databricks:** Managed unified data analytics platform built by original creators of Apache Spark.
