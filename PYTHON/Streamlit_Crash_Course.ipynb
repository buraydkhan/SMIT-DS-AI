## 💡 Module 1: Introduction & Environment Setup

### How Streamlit Works
Unlike traditional web frameworks (Flask, Django, React), Streamlit follows a unique **Reactive Execution Model**:
- Your script runs from **top to bottom** every time a user interacts with a widget.
- You write UI code as if it were a simple linear script.
- Streamlit handles all state synchronization and front-end rendering automatically under the hood.

## 📝 Module 2: Text & Formatting Elements

Streamlit provides intuitive functions for display text, headers, code, math formulas, and general markdown.

### Key Text Functions:
- `st.title("...")`: Main page title
- `st.header("...")`: Section header
- `st.subheader("...")`: Sub-section header
- `st.markdown("...")`: Full GitHub-flavored markdown support
- `st.caption("...")`: Small muted text
- `st.code("...", language="python")`: Code block with syntax highlighting
- `st.latex(r"...")`: Mathematical equations using LaTeX notation
- `st.write(...)`: Streamlit's "Swiss Army knife" function that renders almost any Python object automatically.


````python
%%writefile demo_text.py
import streamlit as st

# Title & Subheaders
st.title("🚀 Streamlit Text Demo")
st.header("1. Header Example")
st.subheader("1.1 Subheader Example")

# Markdown & Formatting
st.markdown("Streamlit supports **bold**, *italic*, [links](https://streamlit.io), and even :sparkles: emojis!")
st.caption("This is a small caption text for footnote or metadata.")

# Code display
code_example = '''
def add_numbers(a, b):
    return a + b
'''
st.code(code_example, language="python")

# LaTeX Math
st.latex(r"E = mc^2 \quad 	ext{and} \quad \sigma = \sqrt{rac{1}{N} \sum_{i=1}^N (x_i - \mu)^2}")

# Divider
st.divider()

# Magic st.write()
st.write("### `st.write()` is super flexible:")
st.write("You can pass strings, dictionaries, numbers:", {"name": "Alice", "role": "Data Scientist", "score": 98.5})

````

## 📊 Module 3: Displaying Data & Media

Streamlit shines when working with DataFrames, tables, JSON objects, metrics, and charts.

### Key Data Display Functions:
- `st.dataframe()`: Interactive DataFrame viewer (sorting, searching, filtering, expanding).
- `st.data_editor()`: Editable data grid where users can edit values directly!
- `st.table()`: Static HTML table.
- `st.metric(label, value, delta)`: KPI display card with percent change / trend indicators.
- `st.json()`: Formatted JSON tree viewer.
- **Charts**: `st.line_chart()`, `st.bar_chart()`, `st.scatter_chart()`, `st.pyplot()`, `st.plotly_chart()`.
- **Media**: `st.image()`, `st.audio()`, `st.video()`.



````python
%%writefile demo_data.py
import streamlit as st
import pandas as pd
import numpy as np
import plotly.express as px

st.title("📊 Data & Charts Display")

# Create sample dataset
np.random.seed(42)
df = pd.DataFrame({
    "Product": ["Gadget A", "Gadget B", "Gadget C", "Gadget D"],
    "Sales ($)": [12500, 23400, 18900, 31000],
    "Growth (%)": [12.5, -4.2, 8.7, 15.3]
})

# Display Metrics
st.subheader("1. Key Metrics")
col1, col2, col3 = st.columns(3)
col1.metric("Total Revenue", "$85,800", "+12.4%")
col2.metric("Active Users", "1,420", "+85")
col3.metric("Churn Rate", "2.1%", "-0.4%", delta_color="inverse")

# Interactive DataFrame
st.subheader("2. Interactive Data Table (`st.dataframe`)")
st.dataframe(df, use_container_width=True)

# Editable DataFrame
st.subheader("3. Editable Table (`st.data_editor`)")
edited_df = st.data_editor(df, num_rows="dynamic")

# Charts
st.subheader("4. Built-in Line Chart")
chart_data = pd.DataFrame(np.random.randn(20, 3), columns=['Apples', 'Oranges', 'Bananas'])
st.line_chart(chart_data)

# Interactive Plotly Chart
st.subheader("5. Plotly Integration (`st.plotly_chart`)")
fig = px.bar(df, x="Product", y="Sales ($)", color="Product", title="Sales by Product")
st.plotly_chart(fig, use_container_width=True)

````

## 🎛️ Module 4: Interactive Input Widgets

Widgets allow users to interact with your app. When a user interacts with any widget, Streamlit automatically re-executes the script with the new widget values!

### Summary of Popular Widgets:
| Category | Function | Returns |
| :--- | :--- | :--- |
| **Action** | `st.button("Click Me")` | `True` when clicked |
| | `st.download_button("Download", data)` | `True` when downloaded |
| **Selection** | `st.checkbox("Check option")` | `True` / `False` |
| | `st.radio("Choose", ["A", "B"])` | Selected string |
| | `st.selectbox("Select", options)` | Selected option |
| | `st.multiselect("Select multiple", options)`| List of selected options |
| **Numeric** | `st.slider("Range", min, max)` | Integer / Float |
| | `st.number_input("Enter number")` | Integer / Float |
| **Text** | `st.text_input("Enter text")` | String |
| | `st.text_area("Long description")` | String |
| **Dates/Files**| `st.date_input("Pick date")` | `datetime.date` |
| | `st.file_uploader("Upload CSV")` | `UploadedFile` object |



````python
%%writefile demo_widgets.py
import streamlit as st
import pandas as pd

st.title("🎛️ Interactive Widgets Demo")

# Text & Number Inputs
name = st.text_input("What is your name?", value="Alice")
age = st.number_input("What is your age?", min_value=1, max_value=120, value=25)

# Selection widgets
role = st.selectbox("Select your role:", ["Data Analyst", "ML Engineer", "Software Developer", "Product Manager"])
skills = st.multiselect("Select your skills:", ["Python", "SQL", "Streamlit", "PyTorch", "Docker", "AWS"])

# Slider & Radio
satisfaction = st.slider("Rate your Streamlit experience (1 to 10):", min_value=1, max_value=10, value=9)
theme = st.radio("Preferred Color Mode:", ["Light Mode", "Dark Mode", "System Default"])

# File Uploader
st.subheader("📁 File Uploader Example")
uploaded_file = st.file_uploader("Upload a CSV file", type=["csv"])
if uploaded_file is not None:
    data = pd.read_csv(uploaded_file)
    st.success("File uploaded successfully!")
    st.dataframe(data.head())

# Summary Output
st.divider()
st.subheader("Summary of Inputs:")
st.write(f"Hello **{name}** ({age} yrs)! You are a **{role}** skilled in **{', '.join(skills)}**.")
st.write(f"Rating: **{satisfaction}/10** | Theme: **{theme}**")

````

## 📐 Module 5: Layouts & App Structure

Streamlit provides tools to structure your app with sidebars, multi-column layouts, tabbed navigation, expanders, and temporary status popups.

### Layout Controls:
1. `st.sidebar`: Place controls in a collapsable left panel.
2. `st.columns([w1, w2, ...])`: Divide main area into flexible side-by-side columns.
3. `st.tabs(["Tab 1", "Tab 2"])`: Group content into tabbed panels.
4. `st.expander("Title")`: Create expandable accordion sections.
5. `st.container()`: Group elements together.
6. `st.empty()`: Single-element container to dynamically overwrite content (useful for timers/animations).
7. **Status elements**: `st.spinner()`, `st.toast()`, `st.balloons()`, `st.snow()`, `st.success()`, `st.error()`.



````python
%%writefile demo_layout.py
import streamlit as st
import time

st.title("📐 Layouts & Organization")

# 1. Sidebar Layout
st.sidebar.title("⚙️ Dashboard Controls")
user_mode = st.sidebar.selectbox("Select View Mode", ["Basic", "Advanced", "Settings"])
show_raw_data = st.sidebar.checkbox("Show Raw Data", value=True)

# 2. Multi-column Layout
st.subheader("1. Flexible Columns (`st.columns`)")
col1, col2, col3 = st.columns([2, 1, 1])

with col1:
    st.info("Column 1 (Width ratio: 2)")
    st.write("Main content or larger chart goes here.")

with col2:
    st.success("Column 2 (Width ratio: 1)")
    st.metric("Users", "5,400")

with col3:
    st.warning("Column 3 (Width ratio: 1)")
    st.metric("Latency", "42ms")

# 3. Tabs Layout
st.subheader("2. Tabbed Interfaces (`st.tabs`)")
tab1, tab2, tab3 = st.tabs(["📈 Overview", "📊 Deep Dive", "⚙️ Settings"])

with tab1:
    st.write("### Overview Panel")
    st.write("High-level summaries and key metrics belong here.")

with tab2:
    st.write("### Deep Dive Analysis")
    st.write("Detailed charts and filter controls go here.")

with tab3:
    st.write("### Application Settings")
    st.slider("API Timeout (seconds)", 1, 30, 5)

# 4. Expanders
st.subheader("3. Expanders (`st.expander`)")
with st.expander("❓ Frequently Asked Questions (Click to expand)"):
    st.write("Streamlit is completely free and open-source under Apache 2.0 license.")

# 5. Status & Animations
if st.button("🎉 Celebrate & Trigger Spinner"):
    with st.spinner("Processing request..."):
        time.sleep(1.5)
    st.balloons()
    st.toast("Process completed successfully!", icon="✅")

````

## 🔄 Module 6: Execution Model & State Management (`st.session_state`)

### The Streamlit Execution Loop:
1. When user opens app: Script runs top to bottom.
2. When user moves a slider / clicks a button: **Script re-runs completely top to bottom**.
3. **Problem**: Any regular Python variable (e.g. `count = 0`) gets reset back to `0` on every re-run!

### The Solution: `st.session_state`
`st.session_state` is a dictionary-like object that persists state across re-runs for a specific user session.

### Example: Counter App
Let's see how `st.session_state` preserves counter values.


````python
%%writefile demo_state.py
import streamlit as st

st.title("🔄 Session State Counter Demo")

# Step 1: Initialize session state variable if it doesn't exist
if 'counter' not in st.session_state:
    st.session_state.counter = 0

if 'user_name' not in st.session_state:
    st.session_state.user_name = "Guest"

# Callback functions
def increment_counter():
    st.session_state.counter += 1

def decrement_counter():
    st.session_state.counter -= 1

def reset_counter():
    st.session_state.counter = 0

# UI Controls using callbacks
st.subheader(f"Current Count: `{st.session_state.counter}`")

col1, col2, col3 = st.columns(3)
col1.button("➕ Increment", on_click=increment_counter)
col2.button("➖ Decrement", on_click=decrement_counter)
col3.button("🔄 Reset", on_click=reset_counter)

st.divider()

# Storing user inputs in session state
name_input = st.text_input("Update Session User Name:", value=st.session_state.user_name)
if st.button("Save Name"):
    st.session_state.user_name = name_input
    st.rerun()

st.write(f"Logged-in User: **{st.session_state.user_name}**")

````

## 📝 Module 7: Forms & Batching Inputs (`st.form`)

### Why use `st.form`?
By default, every widget interaction immediately triggers a page re-run. If you have 5 text input boxes in a form, Streamlit would re-run 5 times as the user types!

`st.form` batches all widgets inside it. The app only re-runs when the user clicks the `st.form_submit_button()`.



````python
%%writefile demo_form.py
import streamlit as st

st.title("📝 Forms & Batch Submission")

st.markdown("With `st.form`, inputs won't trigger re-runs until **Submit** is pressed!")

with st.form(key="user_registration_form"):
    st.subheader("User Registration")
    
    col1, col2 = st.columns(2)
    with col1:
        first_name = st.text_input("First Name")
        email = st.text_input("Email Address")
    with col2:
        last_name = st.text_input("Last Name")
        password = st.text_input("Password", type="password")
        
    subscription = st.selectbox("Plan Tier", ["Free", "Pro", "Enterprise"])
    agree = st.checkbox("I accept the Terms & Conditions")
    
    # Submit button MUST be inside the form
    submit_button = st.form_submit_button(label="Submit Registration")

if submit_button:
    if not agree:
        st.error("You must agree to the Terms & Conditions.")
    elif not first_name or not email:
        st.warning("Please fill out all required fields.")
    else:
        st.success(f"Welcome aboard, {first_name}! Account created on the **{subscription}** plan.")

````

## ⚡ Module 8: Caching & Performance Optimization

Since Streamlit re-runs scripts on user interactions, slow operations like loading large CSVs, querying databases, or running ML models will freeze the app if executed every time.

Streamlit provides two powerful decorator functions for caching:

1. **`@st.cache_data`** (Recommended for Data & Calculations):
   - Caches return values of functions (DataFrames, API responses, query outputs).
   - Creates a fresh copy for each caller so mutation is safe.
   - Use options like `ttl=3600` (time-to-live in seconds).

2. **`@st.cache_resource`** (Recommended for Heavy Shared Objects):
   - Caches global resources (Machine Learning models, PyTorch/TensorFlow weights, Database connection pools).
   - Returns exact singleton reference across all sessions.



````python
%%writefile demo_caching.py
import streamlit as st
import pandas as pd
import numpy as np
import time

st.title("⚡ Streamlit Caching Demo")

# 1. Cache Data Example
@st.cache_data(ttl=600, show_spinner="Loading large dataset...")
def load_large_dataframe(rows):
    time.sleep(2)  # Simulate slow network/CSV load
    df = pd.DataFrame({
        "ID": range(1, rows + 1),
        "Score": np.random.randn(rows),
        "Category": np.random.choice(["Alpha", "Beta", "Gamma"], size=rows)
    })
    return df

# 2. Cache Resource Example (e.g. ML Model)
@st.cache_resource
def load_ml_model():
    time.sleep(3)  # Simulate loading ML model weights
    return {"model_name": "ResNet-50", "status": "Ready", "weights_loaded": True}

st.subheader("1. Testing @st.cache_data")
st.write("Clicking the button below loads data. **1st time takes ~2s, subsequent clicks are instant!**")

if st.button("Load Dataset"):
    start_time = time.time()
    data = load_large_dataframe(50000)
    elapsed = time.time() - start_time
    st.success(f"Data loaded in {elapsed:.4f} seconds!")
    st.dataframe(data.head())

st.divider()

st.subheader("2. Testing @st.cache_resource")
if st.button("Load ML Model"):
    start_time = time.time()
    model = load_ml_model()
    elapsed = time.time() - start_time
    st.info(f"Model `{model['model_name']}` ready in {elapsed:.4f} seconds!")

````

## 🛠️ Module 9: Hands-On Project — Interactive Analytics Dashboard

Now let's combine everything we've learned into a complete, beautiful production-grade Streamlit application!

Features included:
- Sidebar controls & state filtering
- KPI Summary Metrics
- Multi-tab visualization layout (Plotly charts & Data view)
- CSV Export download button



````python
%%writefile app.py
import streamlit as st
import pandas as pd
import numpy as np
import plotly.express as px

# Page Configuration
st.set_page_config(
    page_title="Executive Sales Dashboard",
    page_icon="📈",
    layout="wide"
)

# ---------------------------------------------------------
# Data Loader with Caching
# ---------------------------------------------------------
@st.cache_data
def generate_data():
    np.random.seed(42)
    dates = pd.date_range(start="2026-01-01", periods=100)
    regions = ["North America", "Europe", "Asia-Pacific", "Latin America"]
    categories = ["Software", "Hardware", "Services"]
    
    data = []
    for d in dates:
        for r in regions:
            for c in categories:
                sales = np.random.randint(500, 5000)
                units = np.random.randint(5, 50)
                data.append({"Date": d, "Region": r, "Category": c, "Sales": sales, "Units": units})
                
    return pd.DataFrame(data)

df_raw = generate_data()

# ---------------------------------------------------------
# Sidebar Filters
# ---------------------------------------------------------
st.sidebar.title("📊 Filter Options")
selected_regions = st.sidebar.multiselect(
    "Select Regions:",
    options=df_raw["Region"].unique(),
    default=df_raw["Region"].unique()
)

selected_categories = st.sidebar.multiselect(
    "Select Categories:",
    options=df_raw["Category"].unique(),
    default=df_raw["Category"].unique()
)

min_date, max_date = df_raw["Date"].min().date(), df_raw["Date"].max().date()
date_range = st.sidebar.date_input("Date Range:", value=(min_date, max_date))

# Filter Dataset
filtered_df = df_raw[
    (df_raw["Region"].isin(selected_regions)) &
    (df_raw["Category"].isin(selected_categories))
]

# ---------------------------------------------------------
# Main App Header
# ---------------------------------------------------------
st.title("📈 Executive Sales & Analytics Dashboard")
st.markdown("Interactive business intelligence portal built with **Streamlit**.")

# KPI Cards
total_sales = filtered_df["Sales"].sum()
total_units = filtered_df["Units"].sum()
avg_order_val = filtered_df["Sales"].mean() if len(filtered_df) > 0 else 0

col1, col2, col3, col4 = st.columns(4)
col1.metric("Total Revenue", f"${total_sales:,.0f}", "+8.5%")
col2.metric("Units Sold", f"{total_units:,}", "+4.2%")
col3.metric("Avg Order Value", f"${avg_order_val:,.2f}", "+2.1%")
col4.metric("Active Regions", f"{len(selected_regions)} / {len(df_raw['Region'].unique())}")

st.divider()

# ---------------------------------------------------------
# Tabs & Visualizations
# ---------------------------------------------------------
tab_analytics, tab_data = st.tabs(["📊 Sales Analytics", "📋 Data Explorer"])

with tab_analytics:
    col_chart1, col_chart2 = st.columns(2)
    
    with col_chart1:
        st.subheader("Sales Trend Over Time")
        time_df = filtered_df.groupby("Date")["Sales"].sum().reset_index()
        fig_line = px.line(time_df, x="Date", y="Sales", title="Daily Revenue ($)")
        st.plotly_chart(fig_line, use_container_width=True)
        
    with col_chart2:
        st.subheader("Revenue by Region & Category")
        fig_bar = px.bar(
            filtered_df, x="Region", y="Sales", color="Category",
            barmode="group", title="Sales by Region"
        )
        st.plotly_chart(fig_bar, use_container_width=True)

with tab_data:
    st.subheader("Raw Data Preview")
    st.dataframe(filtered_df, use_container_width=True)
    
    # Export CSV Button
    csv_data = filtered_df.to_csv(index=False).encode('utf-8')
    st.download_button(
        label="📥 Download Filtered Data as CSV",
        data=csv_data,
        file_name="filtered_sales_data.csv",
        mime="text/csv"
    )

````

## 🚀 Module 10: Running & Deploying Streamlit Apps

### How to Run Streamlit Apps Locally
Open your command terminal (Command Prompt / PowerShell / Bash) in the directory containing `app.py` and run:

```bash
streamlit run app.py
```

Streamlit will automatically launch a web browser window at `http://localhost:8501`.

1. Push your code and `requirements.txt` to a **GitHub repository**.
   - Your `requirements.txt` should list dependencies:
     ```text
     streamlit
     pandas
     plotly
     ```
2. Visit **[share.streamlit.io](https://share.streamlit.io)** and log in with your GitHub account.
3. Click **"New App"**, select your repository, branch (`main`), and target file (`app.py`).
4. Click **"Deploy!"** — Your app will be live on a public URL in seconds! 🎈
