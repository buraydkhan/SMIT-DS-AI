```python
# ==============================================================================
# SEABORN BASICS CRASH COURSE
# ==============================================================================

import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Set global Seaborn theme/style
sns.set_theme(style="whitegrid")

# Load built-in sample dataset for demonstration
tips = sns.load_dataset("tips")

print("=== SAMPLE DATASET (Tips) ===")
print(tips.head())
print("-" * 50)
```

```python
# ------------------------------------------------------------------------------
# 1. CATEGORICAL PLOTS (Barplot, Countplot, Boxplot, Violinplot)
# ------------------------------------------------------------------------------
print("\n=== 1. CATEGORICAL PLOTS ===")

fig, axes = plt.subplots(2, 2, figsize=(12, 8))

# Countplot: Counts total rows per category
sns.countplot(data=tips, x="day", hue="sex", ax=axes[0, 0], palette="Set2")
axes[0, 0].set_title("1. Countplot: Customer Counts by Day & Gender")

# Barplot: Shows average (mean) value with confidence intervals
sns.barplot(data=tips, x="day", y="total_bill", hue="sex", ax=axes[0, 1], palette="Set2")
axes[0, 1].set_title("2. Barplot: Average Total Bill by Day & Gender")

# Boxplot: Distribution & Outliers
sns.boxplot(data=tips, x="day", y="total_bill", ax=axes[1, 0], palette="Pastel1")
axes[1, 0].set_title("3. Boxplot: Bill Distribution across Days")

# Violinplot: Kernel Density Estimate + Quartiles
sns.violinplot(data=tips, x="day", y="total_bill", hue="time", split=True, ax=axes[1, 1], palette="Pastel2")
axes[1, 1].set_title("4. Violinplot: Bill Spread by Day & Time (Lunch/Dinner)")

plt.tight_layout()
print("Displaying Categorical Plots... (Close plot window to continue)")
plt.show()
print("-" * 50)
```

```python
# ------------------------------------------------------------------------------
# 2. RELATIONAL PLOTS (Scatterplot & Regplot)
# ------------------------------------------------------------------------------
print("\n=== 2. RELATIONAL PLOTS ===")

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# Scatterplot: Multidimensional analysis using color (hue) and size
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="day",
    size="size",
    sizes=(20, 200),
    palette="deep",
    ax=ax1
)
ax1.set_title("Scatterplot: Total Bill vs. Tip")

# Regplot: Scatterplot with automatically fitted linear regression line
sns.regplot(
    data=tips,
    x="total_bill",
    y="tip",
    scatter_kws={"alpha": 0.5},
    line_kws={"color": "red"},
    ax=ax2
)
ax2.set_title("Regplot: Bill vs. Tip with Regression Line")

plt.tight_layout()
print("Displaying Relational Plots... (Close plot window to continue)")
plt.show()
print("-" * 50)
```

```python
# ------------------------------------------------------------------------------
# 3. DISTRIBUTION PLOTS (Histplot & Kdeplot)
# ------------------------------------------------------------------------------
print("\n=== 3. DISTRIBUTION PLOTS ===")

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

# Histplot: Histogram with optional Kernel Density Estimation (KDE) curve
sns.histplot(data=tips, x="total_bill", kde=True, color="teal", ax=ax1)
ax1.set_title("Histplot with KDE Curve")

# Kdeplot: Smooth density curve comparison across groups
sns.kdeplot(data=tips, x="total_bill", hue="time", fill=True, common_norm=False, ax=ax2)
ax2.set_title("KDE Density Estimation (Lunch vs Dinner)")

plt.tight_layout()
print("Displaying Distribution Plots... (Close plot window to continue)")
plt.show()
print("-" * 50)
```

```python
# ------------------------------------------------------------------------------
# 4. HEATMAP (Correlation Matrix)
# ------------------------------------------------------------------------------
print("\n=== 4. HEATMAP ===")

# Compute correlation matrix for numeric columns
numeric_tips = tips.select_dtypes(include=[np.number])
corr = numeric_tips.corr()

plt.figure(figsize=(6, 5))
# annot=True shows exact numbers; cmap defines the color gradient
sns.heatmap(corr, annot=True, cmap="coolwarm", vmin=-1, vmax=1, linewidths=0.5)
plt.title("Correlation Heatmap")

print("Displaying Heatmap... (Close plot window to continue)")
plt.show()
print("-" * 50)
```

```python
# ------------------------------------------------------------------------------
# 5. PAIRPLOT (Multi-Plot Grid)
# ------------------------------------------------------------------------------
print("\n=== 5. PAIRPLOT ===")

# Generates pairwise scatter plots for all numerical columns automatically
pair_grid = sns.pairplot(tips, hue="species" if "species" in tips else "sex", palette="husl")
pair_grid.fig.suptitle("Pairwise Relationships in Dataset", y=1.02)

print("Displaying Pairplot... (Close plot window to continue)")
plt.show()

print("-" * 50)
print("\n=== ALL SEABORN CODE EXECUTED SUCCESSFULLY! ===")
```
