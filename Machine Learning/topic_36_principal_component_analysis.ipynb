### Topic 36: Principal Component Analysis (PCA)
**Intuition**: Compresses many features down to 2 or 3 principal component axes while preserving maximum variance.

```python
import numpy as np
from sklearn.decomposition import PCA

X_blobs = np.random.randn(150, 4)
pca = PCA(n_components=2)
X_pca_2d = pca.fit_transform(X_blobs)
print(f"Explained Variance Ratio: {pca.explained_variance_ratio_}")
```
