### Topic 32: ROC Curve & AUC Score Intuition
**Intuition**: Plots True Positive Rate vs False Positive Rate as threshold varies from 0 to 1. **AUC (Area Under Curve)**: 1.0 is perfect, 0.5 is random guessing.

```python
from sklearn.metrics import roc_auc_score

y_true_c = [1, 0, 1, 1, 0, 1, 0, 0]
y_probs = [0.9, 0.1, 0.85, 0.4, 0.2, 0.95, 0.6, 0.05]

print(f"ROC-AUC Score: {roc_auc_score(y_true_c, y_probs):.3f}")
```
