### Topic 34: K-Means Clustering Intuition & Code
**Intuition**: Groups unlabeled data points into $K$ clusters around central centroids.

```python
import numpy as np
from sklearn.cluster import KMeans

X_blobs = np.random.randn(150, 2)
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
cluster_labels = kmeans.fit_predict(X_blobs)
print(f"Cluster Centroids Shape: {kmeans.cluster_centers_.shape}")
```
