### Topic 35: Finding Optimal K (Elbow Method)
Plotting Inertia vs $K$ to spot the "elbow" inflection point.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

X_blobs = np.random.randn(150, 2)
inertias = [KMeans(n_clusters=k, random_state=42, n_init=5).fit(X_blobs).inertia_ for k in range(1, 6)]

plt.plot(range(1, 6), inertias, 'bo-')
plt.title('Elbow Method for K-Means')
plt.xlabel('K')
plt.ylabel('Inertia')
plt.show()
```
