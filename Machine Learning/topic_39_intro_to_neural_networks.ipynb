### Topic 39: Intro to Neural Networks & Perceptrons
**Intuition**: Inspired by biological neurons. Inputs are weighted, summed, passed through activation functions (like ReLU), and outputted.

```python
import numpy as np
from sklearn.neural_network import MLPClassifier

X = np.random.randn(100, 2)
y = np.random.choice([0, 1], 100)

mlp = MLPClassifier(hidden_layer_sizes=(16, 8), max_iter=200, random_state=42)
mlp.fit(X, y)
print(f"Neural Network Accuracy: {mlp.score(X, y):.2%}")
```
