### Topic 15: Lasso Regularization (L1 Penalty & Feature Selection)
**Intuition**: Adds a penalty proportional to *absolute weights* ($\alpha \sum |w|$). Lasso can drive useless feature weights to **exact zero**, performing automatic feature selection!

```python
import numpy as np
from sklearn.linear_model import Lasso

X = np.array([[1, 0.1], [3, 0.2], [5, 0.15]])
y = np.array([10, 30, 50])

lasso = Lasso(alpha=0.5)
lasso.fit(X, y)
print("Lasso Coefficients (Notice zeros):", lasso.coef_)
```
