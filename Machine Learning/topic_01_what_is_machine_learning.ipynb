### Topic 01: What is Machine Learning?
**Intuition**: In traditional programming, humans write explicit `if-else` rules to process data into answers. In Machine Learning, we give the computer data and answers, and it automatically learns the rules!

- **Supervised**: Learning with an answer key (e.g., predicting house prices from features).
- **Unsupervised**: Finding hidden groups without answers (e.g., customer segmentation).
- **Reinforcement**: Learning through trial-and-error rewards (e.g., playing chess).

```python
# Rule-based vs Machine Learning approach
# Goal: Estimate price based on square footage

def predict_price_rules(sqft):
    if sqft < 1000:
        return 150000
    elif sqft < 2000:
        return 300000
    else:
        return 500000

print(f"Rule-based prediction for 1500 sqft: ${predict_price_rules(1500):,}")
```
