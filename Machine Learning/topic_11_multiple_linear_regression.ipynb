### Topic 11: Multiple Linear Regression
**Intuition**: Fits a plane/hyperplane when using multiple input features: $y = w_1 x_1 + w_2 x_2 + \dots + b$.

```python
import numpy as np
from sklearn.linear_model import LinearRegression

X_multi = np.array([[1, 20], [2, 25], [3, 35], [4, 40], [5, 50]])
y_multi = np.array([100, 150, 200, 250, 300])

multi_lr = LinearRegression()
multi_lr.fit(X_multi, y_multi)
print("Feature Weights:", multi_lr.coef_)
```
