### Topic 02: Vectors & Matrices in Machine Learning
**Intuition**: In ML, a **vector** is just a list of numbers representing features of one object (e.g., `[Bedrooms, SqFt, Bathrooms]`). A **matrix** is a table of vectors representing your entire dataset!

**Dot Product**: $\mathbf{x} \cdot \mathbf{w} = x_1 w_1 + x_2 w_2 + \dots + x_n w_n$. It calculates the weighted sum of features.

```python
import numpy as np

# Feature vector (x) and Model weights (w)
features_x = np.array([3.0, 1200.0, 2.0])  # [Bedrooms, SqFt, Bathrooms]
weights_w = np.array([25000.0, 150.0, 15000.0])  # Value assigned to each feature

predicted_price = np.dot(features_x, weights_w)
print(f"Dot product predicted house price: ${predicted_price:,.2f}")
```
