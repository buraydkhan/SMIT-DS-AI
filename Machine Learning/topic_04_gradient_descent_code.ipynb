### Topic 04: Gradient Descent Code Example
**Mathematical Rule**: $w_{new} = w_{old} - \alpha \cdot \frac{dJ}{dw}$

Let us run 10 steps of Gradient Descent to find the optimal weight $w=3$.

```python
import numpy as np

w_current = -1.0  # Starting guess
alpha = 0.2       # Learning rate

for step in range(1, 11):
    gradient = 2 * (w_current - 3)  # Derivative of (w-3)^2 + 5
    w_current = w_current - alpha * gradient
    current_cost = (w_current - 3)**2 + 5
    print(f"Step {step:2d}: Weight = {w_current:6.3f}, Cost = {current_cost:6.3f}")
```
