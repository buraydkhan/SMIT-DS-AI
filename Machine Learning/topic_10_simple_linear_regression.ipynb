### Topic 10: Simple Linear Regression
**Intuition**: Fits a straight line $y = m x + c$ through data points to predict continuous numbers.

```python
import numpy as np
from sklearn.linear_model import LinearRegression

X_reg = np.array([[1], [2], [3], [4], [5]])
y_reg = np.array([2.1, 3.9, 6.1, 8.2, 9.9])

lr_model = LinearRegression()
lr_model.fit(X_reg, y_reg)

print(f"Slope (m): {lr_model.coef_[0]:.2f}")
print(f"Intercept (c): {lr_model.intercept_:.2f}")
print(f"Prediction for X=6: {lr_model.predict([[6]])[0]:.2f}")
```
