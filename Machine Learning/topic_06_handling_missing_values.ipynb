### Topic 06: Handling Missing Values
**Intuition**: Real-world data often has blank/missing entries (`NaN`). **Imputation** fills blanks using strategies like column median or mean.

```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer

df_missing = pd.DataFrame({'Age': [25, 30, np.nan, 45, np.nan, 50]})

imputer = SimpleImputer(strategy='median')
df_imputed = pd.DataFrame(imputer.fit_transform(df_missing), columns=['Age'])

print("Before Imputation:\n", df_missing)
print("\nAfter Imputation (Median):\n", df_imputed)
```
