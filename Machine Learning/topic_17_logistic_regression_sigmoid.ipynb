### Topic 17: Logistic Regression & Sigmoid Function
**Intuition**: Converts linear outputs into probabilities between 0 and 1 using the Sigmoid curve: $\sigma(z) = \frac{1}{1 + e^{-z}}$.

```python
import numpy as np
import matplotlib.pyplot as plt

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

z_vals = np.linspace(-6, 6, 100)
plt.figure(figsize=(6, 3.5))
plt.plot(z_vals, sigmoid(z_vals), color='teal', lw=2)
plt.axhline(0.5, color='gray', linestyle='--')
plt.title('Sigmoid Function Curve')
plt.xlabel('Linear Output (z)')
plt.ylabel('Probability')
plt.show()
```
