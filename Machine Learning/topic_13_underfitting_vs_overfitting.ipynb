### Topic 13: Underfitting vs. Overfitting
**Intuition**:
- **Underfitting (High Bias)**: Model is too simple (e.g. straight line for curvy data).
- **Overfitting (High Variance)**: Model is too complex and memorizes noise.

```python
import numpy as np
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

X_curve = np.linspace(-3, 3, 30).reshape(-1, 1)
y_curve = X_curve.ravel()**2 + np.random.normal(0, 1, 30)

poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X_curve)
poly_model = LinearRegression().fit(X_poly, y_curve)

print(f"Polynomial R2 Score: {poly_model.score(X_poly, y_curve):.3f}")
```
