### Topic 05: Loading & Exploring Data (EDA)
**Intuition**: EDA is like inspecting your raw ingredients before cooking. We inspect dataset shape, column types, and distributions.

```python
import pandas as pd
import numpy as np

np.random.seed(42)
df = pd.DataFrame({
    'Age': np.random.randint(18, 65, 100),
    'Salary': np.random.normal(55000, 15000, 100),
    'City': np.random.choice(['NYC', 'SF', 'Chicago'], 100)
})

print(df.head())
print("\nSummary Statistics:")
print(df.describe())
```
