### Topic 21: Support Vector Machines (SVM) - Maximum Margin
**Intuition**: Finds a decision boundary line that maximizes the gap/margin between the two closest classes (Support Vectors).

```python
from sklearn.svm import SVC
from sklearn.datasets import make_classification

X_clf, y_clf = make_classification(n_samples=100, n_features=2, n_redundant=0, random_state=42)

svm_linear = SVC(kernel='linear')
svm_linear.fit(X_clf, y_clf)
print(f"Number of Support Vectors: {len(svm_linear.support_vectors_)}")
```
