### Topic 03: Cost Functions & Gradient Descent Intuition
**Intuition**: Imagine you are at the top of a foggy mountain and want to reach the valley (lowest error). **Gradient Descent** looks at the slope beneath your feet and takes steps downhill!

- **Cost Function $J(w)$**: Measures how wrong the model is.
- **Learning Rate $\alpha$**: Step size. Too large = overshooting; too small = super slow.

```python
import numpy as np
import matplotlib.pyplot as plt

# Cost function J(w) = (w - 3)^2 + 5
w = np.linspace(-1, 7, 100)
cost = (w - 3)**2 + 5

plt.figure(figsize=(7, 4))
plt.plot(w, cost, label='Cost Function J(w)', color='indigo', lw=2)
plt.scatter([3], [5], color='red', s=100, zorder=5, label='Minimum Error (Valley)')
plt.title('Cost Function & The Minimum Error Valley')
plt.xlabel('Weight (w)')
plt.ylabel('Cost J(w)')
plt.legend()
plt.show()
```
