### Topic 19: K-Nearest Neighbors (KNN) Intuition
**Intuition**: "Tell me who your neighbors are, and I will tell you who you are." Predicts class based on majority vote of $K$ closest points.

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.datasets import make_classification

X_clf, y_clf = make_classification(n_samples=100, n_features=2, n_redundant=0, random_state=42)

knn_3 = KNeighborsClassifier(n_neighbors=3)
knn_3.fit(X_clf, y_clf)
print(f"KNN (K=3) Accuracy: {knn_3.score(X_clf, y_clf):.2%}")
```
