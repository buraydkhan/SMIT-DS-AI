### Topic 16: What is Classification?
**Intuition**: Predicts discrete class categories (e.g. Spam vs Not Spam, Cat vs Dog vs Bird).

```python
from sklearn.datasets import make_classification
import numpy as np

X_clf, y_clf = make_classification(n_samples=100, n_features=2, n_redundant=0, random_state=42)
print(f"X shape: {X_clf.shape} | Unique classes: {np.unique(y_clf)}")
```
