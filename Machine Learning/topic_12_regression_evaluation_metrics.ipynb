### Topic 12: Regression Evaluation Metrics
**Intuition**:
- **MAE (Mean Absolute Error)**: Average error in original units.
- **MSE (Mean Squared Error)**: Penalizes large errors heavily.
- **R² Score**: Percentage of variance explained (1.0 = perfect prediction).

```python
import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

y_true = np.array([3.0, -0.5, 2.0, 7.0])
y_pred = np.array([2.5, 0.0, 2.1, 7.8])

print(f"MAE : {mean_absolute_error(y_true, y_pred):.3f}")
print(f"MSE : {mean_squared_error(y_true, y_pred):.3f}")
print(f"R2  : {r2_score(y_true, y_pred):.3f}")
```
