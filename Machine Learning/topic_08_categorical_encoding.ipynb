### Topic 08: Categorical Encoding (One-Hot Encoding)
**Intuition**: ML models only understand numbers, not text like `'NYC'` or `'SF'`. **One-Hot Encoding** creates binary (0 or 1) columns for each category.

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

df = pd.DataFrame({'City': ['NYC', 'SF', 'Chicago', 'NYC', 'SF']})

encoder = OneHotEncoder(sparse_output=False)
city_encoded = encoder.fit_transform(df[['City']])
encoded_df = pd.DataFrame(city_encoded, columns=encoder.get_feature_names_out(['City']))
print(encoded_df)
```
