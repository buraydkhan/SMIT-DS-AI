### Topic 14: Ridge Regularization (L2 Penalty)
**Intuition**: Adds a penalty proportional to the *square of weights* ($\alpha \sum w^2$) to keep weights small and prevent overfitting.

```python
import numpy as np
from sklearn.linear_model import Ridge

X = np.array([[1, 2], [3, 4], [5, 6]])
y = np.array([10, 20, 31])

ridge = Ridge(alpha=10.0)
ridge.fit(X, y)
print("Ridge Coefficients:", ridge.coef_)
```
