### Topic 07: Feature Scaling (Standardization)
**Intuition**: If `Age` is around 30 and `Salary` is around 60,000, algorithms will think `Salary` is 2000x more important just because the numbers are bigger! Scaling puts all features on equal footing.

**Standardization**: $x' = \frac{x - \mu}{\sigma}$ (Mean = 0, Std = 1)

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler

df = pd.DataFrame({
    'Age': [25, 35, 45, 55],
    'Salary': [45000, 65000, 85000, 105000]
})

scaler = StandardScaler()
df_scaled = pd.DataFrame(scaler.fit_transform(df), columns=['Age_Scaled', 'Salary_Scaled'])
print(df_scaled)
```
