### Topic 30: Confusion Matrix & Accuracy
**Intuition**:
- **TP (True Positive)**: Correctly predicted positive.
- **TN (True Negative)**: Correctly predicted negative.
- **FP (False Positive - Type I Error)**: False alarm.
- **FN (False Negative - Type II Error)**: Missed detection.

```python
from sklearn.metrics import confusion_matrix

y_true_c = [1, 0, 1, 1, 0, 1, 0, 0]
y_pred_c = [1, 0, 1, 0, 0, 1, 1, 0]

cm = confusion_matrix(y_true_c, y_pred_c)
print("Confusion Matrix:\n", cm)
```
