### Topic 40: Multi-Layer Perceptron (MLP) Loss Curve
Visualizing how loss decreases over training iterations in a Neural Network.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.neural_network import MLPClassifier

X = np.random.randn(100, 2)
y = np.random.choice([0, 1], 100)

mlp = MLPClassifier(hidden_layer_sizes=(16, 8), max_iter=200, random_state=42)
mlp.fit(X, y)

plt.figure(figsize=(6, 3.5))
plt.plot(mlp.loss_curve_, color='crimson', lw=2)
plt.title('MLP Neural Network Training Loss Curve')
plt.xlabel('Iterations')
plt.ylabel('Loss')
plt.show()
```
