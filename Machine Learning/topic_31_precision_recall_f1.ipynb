### Topic 31: Precision, Recall & F1-Score Intuition
**Intuition**:
- **Precision** = $\frac{TP}{TP + FP}$: When model predicts Spam, how often is it right?
- **Recall** = $\frac{TP}{TP + FN}$: Out of all real Spam emails, how many did we catch?
- **F1-Score**: Harmonic mean balancing Precision and Recall.

```python
from sklearn.metrics import precision_score, recall_score, f1_score

y_true_c = [1, 0, 1, 1, 0, 1, 0, 0]
y_pred_c = [1, 0, 1, 0, 0, 1, 1, 0]

print(f"Precision: {precision_score(y_true_c, y_pred_c):.2f}")
print(f"Recall   : {recall_score(y_true_c, y_pred_c):.2f}")
print(f"F1-Score : {f1_score(y_true_c, y_pred_c):.2f}")
```
